4. Imprime nuevamente las cantidad de valores únicos por columna.
- ¿Por qué son 144 horas diferentes?
- La cantidad de combinaciones únicas de fecha y hora no coincide con el número total de mediciones. Aunque no haya valores NaN, esto no garantiza que no se hayan omitido registros. Identifica las fechas en las que faltan mediciones.
- Una vez concluido el análisis de mediciones faltantes, elimina todos los registros que no correspondan al año 2020 y la columna `medicion`.


```python
valores_unicos_actualizados = weather_df.nunique().to_frame("valores_unicos")
combinaciones_observadas = weather_df[["fecha", "hora"]].drop_duplicates().shape[0]
combinaciones_posibles = weather_df["fecha"].nunique() * weather_df["hora"].nunique()
mediciones_por_fecha = weather_df.groupby("fecha").size().rename("mediciones")
fechas_con_conteo_incompleto = mediciones_por_fecha[mediciones_por_fecha.ne(144)]

rango_esperado = pd.date_range(
    start=weather_df["medicion"].min(),
    end=weather_df["medicion"].max(),
    freq="10min"
)
mediciones_faltantes = rango_esperado.difference(
    pd.DatetimeIndex(weather_df["medicion"])
)
faltantes_df = pd.DataFrame({"medicion_faltante": mediciones_faltantes})
faltantes_df["fecha"] = faltantes_df["medicion_faltante"].dt.date
faltantes_df["hora"] = faltantes_df["medicion_faltante"].dt.time

display(valores_unicos_actualizados)
print(f"Combinaciones observadas de fecha y hora: {combinaciones_observadas}")
print(f"Combinaciones posibles por producto cartesiano: {combinaciones_posibles}")
display(fechas_con_conteo_incompleto.to_frame())
display(faltantes_df)

weather_df = (
    weather_df.loc[weather_df["medicion"].dt.year.eq(2020)]
    .drop(columns="medicion")
    .reset_index(drop=True)
)

print(f"Dimensionalidad final para 2020: {weather_df.shape}")
display(weather_df.head())
```

**Respuesta.** Hay **144 horas distintas** porque el día tiene 24 horas y se registra una observación cada 10 minutos: $24\times 6=144$ intervalos diarios. El conteo por fecha muestra 143 registros el 1 de enero de 2020, 135 el 29 de mayo y uno el 1 de enero de 2021. Los extremos de enero se explican por el límite temporal del extracto: inicia a las 00:10 del 1 de enero de 2020 y termina a las 00:00 del 1 de enero de 2021.

Al construir el rango continuo esperado entre la primera y la última medición, la única omisión real aparece el **29 de mayo de 2020**, con nueve observaciones ausentes entre **09:40 y 11:00**. Finalmente se conserva únicamente el año 2020 y se elimina `medicion`; el dataframe resultante tiene **52,694 filas y 14 columnas**.

5. Obtén un dataframe `indicadores_diarios` que sólo almacene el mayor valor de las mediciones por fecha para responder:

8. Utiliza el dataframe de indicadores diarios para crear un nuevo dataframe `indicadores_mensuales`. Para ello:
- Crea una nueva columna `mes` a partir de la fecha que se encuentra en el índice.
- Haciendo uso de `groupby()`, calcula los promedios mensuales y conserva únicamente las columnas `humedad_relativa`, `temperatura_celsius` y `velocidad_viento` en el nuevo dataframe.

```python
indicadores_diarios_con_mes = indicadores_diarios.copy()
indicadores_diarios_con_mes["mes"] = pd.to_datetime(
    indicadores_diarios_con_mes.index
).month

columnas_mensuales = [
    "humedad_relativa", "temperatura_celsius", "velocidad_viento"
]

indicadores_mensuales = (
    indicadores_diarios_con_mes.groupby("mes")[columnas_mensuales]
    .mean()
)

display(indicadores_mensuales.round(3))
```

**Respuesta.** El dataframe mensual contiene doce filas, una por mes, y los promedios de los máximos diarios de humedad relativa, temperatura y velocidad de viento. La temperatura alcanza su promedio mensual más alto en agosto, mientras que la velocidad del viento sobresale en febrero.

9. Construye el mismo dataframe mensual, pero esta vez utilizando la función `pivot_table()`.
- Añade la columna `indice_calor` usando la fórmula de Thom (*Temperature - Humidity Index*, TDI):
> `temperatura_celsius + 0.33 x humedad_relativa - 0.7 x velocidad_viento - 4`

```python
base_pivot = indicadores_diarios.reset_index()
base_pivot["mes"] = pd.to_datetime(base_pivot["fecha"]).dt.month

indicadores_mensuales_pivot = base_pivot.pivot_table(
    index="mes",
    values=columnas_mensuales,
    aggfunc="mean"
)[columnas_mensuales]

misma_estructura = indicadores_mensuales_pivot.equals(indicadores_mensuales)

indicadores_mensuales_pivot["indice_calor"] = (
    indicadores_mensuales_pivot["temperatura_celsius"]
    + 0.33 * indicadores_mensuales_pivot["humedad_relativa"]
    - 0.7 * indicadores_mensuales_pivot["velocidad_viento"]
    - 4
)

print(f"El resultado de pivot_table coincide con groupby: {misma_estructura}")
display(indicadores_mensuales_pivot.round(3))
```

**Respuesta.** `pivot_table()` reproduce exactamente los promedios obtenidos mediante `groupby()`. Al aplicar la fórmula indicada, el índice de calor más alto corresponde a **agosto (48.930)**, seguido de septiembre y junio. Este resultado debe interpretarse como el índice definido por la actividad sobre promedios mensuales de máximos diarios, no como una medición fisiológica instantánea.

10. Recuerda que de las cuatro funciones estudiadas para manipular la estructura del dataframe:
- `melt/pivot` hacen las transformaciones de manera controlada, es decir puedes
definir por medio de sus parámetros qué variables quedarán como índices, columnas y valores.
- `stack/unstack` la conversión se aplica siempre sobre los niveles inferiores de index/columns.
- Convierte el dataframe del ejercicio anterior a formato largo usando: `melt()` y `stack()` y comenta las diferencias.

```python
formato_largo_melt = (
    indicadores_mensuales_pivot.reset_index()
    .melt(id_vars="mes", var_name="indicador", value_name="valor")
)

formato_largo_stack = (
    indicadores_mensuales_pivot.stack()
    .rename("valor")
    .reset_index()
    .rename(columns={"level_1": "indicador"})
)

mismos_valores = (
    formato_largo_melt.sort_values(["mes", "indicador"]).reset_index(drop=True)
    .equals(
        formato_largo_stack.sort_values(["mes", "indicador"]).reset_index(drop=True)
    )
)

print(f"Dimensión con melt: {formato_largo_melt.shape}")
print(f"Dimensión con stack: {formato_largo_stack.shape}")
print(f"Ambos contienen los mismos valores: {mismos_valores}")
display(formato_largo_melt.head(8))
display(formato_largo_stack.head(8))
```

**Respuesta.** Ambos métodos generan **48 observaciones** —12 meses por cuatro indicadores— y contienen los mismos datos. `melt()` permite especificar de forma explícita la variable identificadora (`mes`), los nombres de las variables y la columna de valores; por ello resulta más controlable cuando existen columnas que deben conservarse. `stack()` opera directamente sobre el nivel más interno de las columnas y lo desplaza al índice, por lo que es más conciso cuando la estructura indexada ya está organizada. La diferencia visible es el orden: `melt()` agrupa primero por indicador, mientras que `stack()` recorre primero cada mes.
